```python
#Write a PYTHON program to sum the given sequence
n = int(input("Enter the value of n: "))

sum = 1
fact = 1

for i in range(1, n + 1):
    fact = fact * i
    sum = sum + (1 / fact)

print("Sum =", sum)
```

```python
#Write a PYTHON program to compute the cosine series cos(x) = 1 – x2 / 2! + x4 / 4! – x6 / 6! + … xn / n!
x = float(input("Enter the value of x: "))
n = int(input("Enter the highest even power: "))

sum = 1
sign = -1

for i in range(2, n + 1, 2):
    fact = 1
    for j in range(1, i + 1):
        fact = fact * j
    sum = sum + sign * (x ** i) / fact
    sign = sign * -1

print("Cosine series =", sum)
```

```python
#Write a short PYTHON program to check weather the square root of number is prime or  not.
n=int(input("enter the num"))
square=int(n**0.5)
for i in range(2,square):
    if square%i!=0:
        print("Number is not prime")
        break;
else:
    print("not prime")
```
