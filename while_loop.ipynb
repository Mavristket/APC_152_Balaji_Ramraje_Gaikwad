```python
#Print Natural Numbers in Reverse Order
n = int(input("Enter n: "))

while n >= 1:
    print(n, end=" ")
    n -= 1
```

```python
#Fibonacci Series up to n Terms
n = int(input("Enter number of terms: "))

a = 0
b = 1
count = 0

while count < n:
    print(a, end=" ")
    c = a + b
    a = b
    b = c
    count += 1
```

```python
#Factorial of a Number
n = int(input("Enter a number: "))

fact = 1
i = 1

while i <= n:
    fact *= i
    i += 1

print("Factorial =", fact)
```

```python
#Check Prime Number
n = int(input("Enter a number: "))

if n < 2:
    print("Not Prime")
else:
    i = 2
    prime = True

    while i <= n // 2:
        if n % i == 0:
            prime = False
            break
        i += 1

    if prime:
        print("Prime Number")
    else:
        print("Not Prime")
```

```python
#Sum of Digits
n = int(input("Enter a number: "))

sum = 0

while n > 0:
    digit = n % 10
    sum += digit
    n = n // 10

print("Sum of digits =", sum)
```

```python
#Check Palindrome Number
n = int(input("Enter a number: "))

temp = n
rev = 0

while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp = temp // 10

if n == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Reverse a Number
n = int(input("Enter a number: "))

rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print("Reverse =", rev)
```
