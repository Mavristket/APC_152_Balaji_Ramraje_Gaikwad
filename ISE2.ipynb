```python
marks=[]
n=int(input("Enter the number"))
for i in range(n):
    marks.append(int(input()))
average=sum(marks)/len(marks)
count=0
for i in marks:
    if i>average:
        count=count+1
print("The minimum is",min(marks))
print("The maximun is",max(marks))
print("The avrage is",average)
```

```python
number=int(input("Enter the number to reverse"))
reverse=""
while number>0:
    digit=number%10
    reverse=reverse+str(digit)
    number=number//10
print(reverse)
```
